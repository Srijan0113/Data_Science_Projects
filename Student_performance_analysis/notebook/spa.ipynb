```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use("default")
```

```python
df=pd.read_csv(r"C:\Users\Acer\Desktop\data_science_project\student_performance_analysis\data\student_data.csv")
df.head()
df.shape
df.info
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(numeric_only=True),inplace=True)
df["gender"]=df["gender"].fillna(df["gender"].mode()[0])
```

```python
df["average_score"]=df[["math_score","reading_score","writing_score"]].mean(axis=1)
df
```

```python
df["average_score"].describe()
```

```python
df.groupby("gender")["average_score"].mean()
```

```python
sns.barplot(x="gender",y="average_score",data=df)
plt.title("Average score by Gender")
plt.show()
```

```python
#attendance vs performance . How strongly they are related to each other
df["attendance"].corr(df["average_score"])
```

```python
sns.scatterplot(x="attendance",y="average_score",data=df)
plt.title("Attendance vs Average Score")
plt.show
```

```python
df["study_hours"].corr(df["average_score"])
```

```python
sns.regplot(x="study_hours",y="average_score",data=df)
plt.title("Study hours vs Average Score")
plt.show
```

Student Performance Analysis
Analyzed student performance data using Python and Pandas; performed data cleaning, feature engineering, and exploratory data analysis to extract academic insights which are:
1.Students with higher attendance tend to score better academically.
2.Study hours show a positive correlation with average score.
3.Gender-wise performance variation is minimal.
4.Average score is a better performance indicator than individual subject marks.
