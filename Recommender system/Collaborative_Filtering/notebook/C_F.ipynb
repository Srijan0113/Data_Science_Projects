```python
import pandas as pd
import numpy as np
movies_df=pd.read_csv(r'C:\Users\Acer\Desktop\Data_Science_Projects\Recommender system\Collaborative_Filtering\data\movies.csv')
ratings_df=pd.read_csv(r'C:\Users\Acer\Desktop\Data_Science_Projects\Recommender system\Collaborative_Filtering\data\ratings.csv')
movies_df.head()
```

```python
#select a user
import random
userId=random.choice(ratings_df['userId'].unique())
print("Selected User :",userId)
```

```python
userRatings=ratings_df[ratings_df['userId']==userId]
userRatings.head()
```

```python
#create user-movie matrix
userMovieMatrix=ratings_df.pivot_table(index='userId',
                                       columns='movieId',
                                       values='rating')
userMovieMatrix.head()
```

```python
#Fill missing values
userMovieMatrix=userMovieMatrix.fillna(0)
userMovieMatrix.head()
```

```python
#compute similarity between users
from sklearn.metrics.pairwise import cosine_similarity
userSimilarity=cosine_similarity(userMovieMatrix)
userSimilarity_df=pd.DataFrame(userSimilarity,
index=userMovieMatrix.index,
columns=userMovieMatrix.index)
userSimilarity_df.head()
```

```python
#find similar users
similarUsers=userSimilarity_df[userId].sort_values(ascending=False)
similarUsers.head(10)
```

```python
topUsers=similarUsers.iloc[1:6].index
topUsers
```

```python
#Get movies They liked
topUsersRatings=ratings_df[ratings_df['userId'].isin(topUsers)]
topUsersRatings.head()
```

```python
#Compute wighted recommendation
topUsersRatings=topUsersRatings.merge(similarUsers.rename('similarity'),
                                      left_on='userId',
                                      right_index=True)
topUsersRatings.head()
```

```python
#Calculate weighted score
topUsersRatings['weightedRating']=topUsersRatings['rating']*topUsersRatings['similarity']
```

```python
#Aggregate Scores per movie
recommendation_df=topUsersRatings.groupby('movieId')['weightedRating'].sum().reset_index()
recommendation_df.head()
```

```python
#sort recommendations
recommendation_df=recommendation_df.sort_values(by='weightedRating',ascending=False)
recommendation_df.head(10)
```

```python
#remove already watched movies
watched_movies=userRatings['movieId'].tolist()
```

```python
#remove from recommendation
recommendation_df=recommendation_df[~recommendation_df['movieId'].isin(watched_movies)]
```

```python
#Add movies names
recommendation_df=recommendation_df.merge(movies_df[['movieId','title']],on='movieId')
```

```python
#Sort and show final top 10
final_recommendations=recommendation_df.sort_values(
    by='weightedRating',
    ascending=False
)
final_recommendations[['title','weightedRating']].head(10)
```
