```python
import pandas as pd
import numpy as np
train=pd.read_csv(r"C:\Users\Acer\Desktop\Data_Science_Projects\House_Price_Prediction\Kaggles_data\train.csv")
test=pd.read_csv(r"C:\Users\Acer\Desktop\Data_Science_Projects\House_Price_Prediction\Kaggles_data\test.csv")
train.head()
```

```python
print(train.shape)
print(test.shape)
```

```python
train.info()
```

```python
#Remove outliers
train=train.drop(train[(train['GrLivArea']>4000) 
                 &(train['SalePrice']<300000)].index)
train=train.drop(train[(train['OverallQual']<5) 
                 &(train['SalePrice']>200000)].index)
```

```python
#identify target and features
y=np.log1p(train['SalePrice'])
train=train.drop('SalePrice',axis=1)
```

```python
#combine train+test
full=pd.concat([train,test],axis=0).reset_index(drop=True)
```

# Feature Engineering

```python
#total Square feet
full['TotalSF']=(
    full['TotalBsmtSF']+full['1stFlrSF']+full['2ndFlrSF']
)
```

```python
#Total Bathrooms
full['TotalBath']=(
    full['FullBath']+0.5*full['HalfBath']+full['BsmtFullBath']+0.5*full['BsmtHalfBath']
)
```

```python
#Age features
full['HouseAge']=full['YrSold']-full['YearBuilt']
full['RemodAge']=full['YrSold']-full['YearRemodAdd']
```

```python
#Binary Features
full['HasGarage']=(full['GarageArea']>0).astype(int)
full['HasBsmt']=(full['TotalBsmtSF']>0).astype(int)
full['HasPool']=(full['PoolArea']>0).astype(int)
```

```python
#Porch Area
full['TotalPorch']=(
    full['OpenPorchSF']+
    full['EnclosedPorch']+
    full['3SsnPorch']+
    full['ScreenPorch']
)
```

```python
#Interaction Features
full['OverallScore']=full['OverallQual']*full['OverallCond']
full['QualArea']=full['OverallQual']*full['GrLivArea']
```

```python
print("TotalSF" in full.columns)
```

```python
full['OverallQual_GrLiv']=full['OverallQual']*full['GrLivArea']
full['TotalSF_Qual']=full['TotalSF']*full['OverallQual']
full['GarageScore']=full['GarageArea']*full['GarageCars']
full['BathScore']=full['TotalBath']*full["OverallQual"]
```

```python
#Handle missing values
for col in full.select_dtypes(include=[np.number]).columns:
    full[col]=full[col].fillna(full[col].median())
for col in full.select_dtypes(include=['object']).columns:
    full[col]=full[col].fillna("None")
```

```python
from sklearn.model_selection import KFold,cross_val_score
from scipy.stats import skew
```

```python
#Fix Skewed Numerical Features
numeric_f=full.select_dtypes(include=['float64','int64']).columns
skewed=full[numeric_f].apply(lambda x: skew(x.dropna()))
skewed=skewed[abs(skewed)>0.75]
skewed=skewed.index
full[skewed]=np.log1p(full[skewed])
```

```python
important_cols=[
    'GrLivArea',
    'TotalSF',
    'LotArea',
    'TotalPorch'
]
for col in important_cols:
    full[col + '_log']=np.log1p(full[col])
```

```python
#One Hot Encode
full=pd.get_dummies(full)
```

```python
#Split Back
X=full.iloc[:len(train)]
X_test=full.iloc[len(train):]
```

```python
print(X.shape)
```

```python
from sklearn.feature_selection import VarianceThreshold
selector=VarianceThreshold(threshold=0.01)
X=selector.fit_transform(X)
X_test=selector.transform(X_test)
```

```python
from sklearn.feature_selection import SelectFromModel
from xgboost import XGBRegressor
selector_model=XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=3,
    random_state=42,
    n_jobs=-1
)
selector_model.fit(X,y)
selector=SelectFromModel(selector_model,threshold='median',prefit=True)
X_selected=selector.transform(X)
X_test_selected=selector.transform(X_test)

print("Original features",X.shape[1])
print('Selected features : ',X_selected.shape[1])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold,cross_val_score
from sklearn.metrics import mean_absolute_error
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
```

```python

xgb_final=XGBRegressor(
    n_estimators=3000,
    random_state=42,
    learning_rate=0.01,
    max_depth=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.01,
    reg_lambda=1,
    n_jobs=-1
)
```

```python
scores=cross_val_score(
    xgb_final,X_selected,y,
    scoring='neg_root_mean_squared_error',
    cv=5
)
xgb_rmse=-scores.mean()
print('XGB CV',xgb_rmse)
```

```python
xgb_final.fit(X_selected,y)
test_pred_log=xgb_final.predict(X_test_selected)
test_pred=np.expm1(test_pred_log)
```

```python
#Create submission file
submission = pd.DataFrame({
    'ID':test['Id'],
    'SalePrice':test_pred
})
submission.to_csv("submission4.csv",index=False)
```
