```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use("default")
```

```python
df=pd.read_csv(r"C:\Users\Acer\Desktop\data_science_project\Ipl_Match_data_analysis\data\ipl_matches.csv")
df.head()
df.shape
df.info
```

```python
#count match wins per team
team_wins=df["winner"].value_counts()
team_wins
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=team_wins.index,y=team_wins.values)
plt.title("Most successful IPL teams")
plt.xlabel("Team")
plt.ylabel("Number of wins")
plt.xticks(rotation=45)#xticks means the names - rotates them by 45 deg
plt.show()
```

```python
#Toss winner vs Match winner
df["toss_win_match_win"]=df["toss_winner"]==df["winner"]
df["toss_win_match_win"].value_counts()
```

```python
#toss_win_percentage
toss_win_percentage=df["toss_win_match_win"].mean()*100
toss_win_percentage
```

```python
#toss decision impact(bat vs bowl)
df.groupby("toss_decision")["toss_win_match_win"].mean()*100
```

```python
sns.barplot(x="toss_decision",y="toss_win_match_win",data=df)
plt.title("Impact of Toss Desion on Match Outcome")
plt.ylabel("Win Percentage")
plt.show()
```

```python
#player of the match analysis
player_awards=df["player_of_match"].value_counts()
player_awards
```

```python
player_awards.head(5).plot(kind="bar",figsize=(7,4))
plt.title("Top Players of the Match Awards")
plt.ylabel("Count")
plt.show()
```
